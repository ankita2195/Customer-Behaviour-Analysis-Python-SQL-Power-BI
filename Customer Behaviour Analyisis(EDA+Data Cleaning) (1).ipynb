### Correcting the Category Mapping

```python
# Correct category mapping

correct_mapping = {

    # Clothing
    "T-shirt": "Clothing",
    "Shirt": "Clothing",
    "Shorts": "Clothing",
    "Hoodie": "Clothing",
    "Pants": "Clothing",
    "Socks": "Clothing",
    "Jeans": "Clothing",
    "Blouse": "Clothing",
    "Skirt": "Clothing",
    "Sweater": "Clothing",
    "Dress": "Clothing",

    # Accessories
    "Sunglasses": "Accessories",
    "Gloves": "Accessories",
    "Jewelry": "Accessories",
    "Hat": "Accessories",
    "Handbag": "Accessories",
    "Backpack": "Accessories",
    "Belt": "Accessories",
    "Scarf": "Accessories",
    "Watch": "Accessories",
    "Bag": "Accessories",

    # Electronics
    "Laptop": "Electronics",
    "Phone": "Electronics",
    "Headphones": "Electronics",

    # Footwear
    "Shoes": "Footwear",
    "Sandals": "Footwear",
    "Sneakers": "Footwear",
    "Boots": "Footwear",

    # Outerwear
    "Coat": "Outerwear",
    "Jacket": "Outerwear"
}
```

```python
df["Category"] = df["Item Purchased"].map(correct_mapping)
```

```python
df.groupby(by = "Category")["Item Purchased"].unique().reset_index()
```

### Handling Null Values

```python
df.isnull().sum()
```

#### Null Values in Size Column

```python
df['Size'].unique()
```

```python
df.groupby("Category")['Size'].unique()
```

```python
# Replacing all sizes from the electronics and accessories as sizes are not applicable for them

df.loc[df["Category"] == "Electronics", "Size"] = "Not Applicable"
```

```python
df.loc[df["Category"] == "Accessories", "Size"] = "Free Size"
```

```python
# Sizes in Footwear are in numbers the given sizes are not applicable

df.loc[df["Category"] == "Footwear", "Size"] = "Not Available"
```

```python
# Replacing null values from Clothing with mode

df.loc[(df["Category"] == "Clothing") & (df["Size"].isnull()), "Size"] = df[df["Category"]=="Clothing"]["Size"].mode()[0]
```

```python
df.groupby("Category")['Size'].unique()
```

#### Null Values in Review Rating Column

```python
df["Review Rating"].describe()
```

```python
df.groupby(by = "Item Purchased")["Review Rating"].mean()
```

```python
df["Review Rating"] = df["Review Rating"].fillna(df.groupby(by = "Item Purchased")["Review Rating"].transform('mean'))
```

```python
df.head()
```

```python
df["Review Rating"] = df["Review Rating"].round(2)
```

#### Null Values in Previous Purchases Column

```python
df["Previous Purchases"] = df["Previous Purchases"].fillna(0)
```

#### Null Values in Purchase Amount (USD) Column

```python
df.groupby(by = "Item Purchased")["Purchase Amount (USD)"].mean()
```

```python
df["Purchase Amount (USD)"] = df["Purchase Amount (USD)"].fillna(df.groupby(by = "Item Purchased")["Purchase Amount (USD)"].transform("mean"))
```

```python
from sqlalchemy import create_engine

# MySQL Workbench connection details
username = "root"
password = "root"
host = "localhost"
port = "3306"
database = "customer_behaviour_analysis"

# Create connection string
connection_string = (
    f"mysql+pymysql://{username}:{password}@{host}:{port}/{database}"
)

# Create engine
engine = create_engine(connection_string)

print("✅ Successfully connected to MySQL!")
```
